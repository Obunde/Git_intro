# Calculating Efficiency of Delivery Hubs

## Calculating Hub Efficiency

`calculate_efficiency` measures how well a delivery hub performed against its target:

    efficiency = (actual_deliveries / target_deliveries) * 100

Result is rounded to 2 decimal places for clean output, and returns 0 if `target_deliveries` is 0 (avoids a crash from dividing by zero).

We'll reuse this function to loop through all hubs next.

```python
def calculate_efficiency(actual_deliveries, target_deliveries):
    if target_deliveries == 0:
        return 0
    efficiency = actual_deliveries / target_deliveries * 100
    return round(efficiency, 2)
```

## Classifying Operational Status

`get_operational_status` takes an efficiency percentage and labels it into one of three bands:

- **< 70** → `"critical"` — the hub is significantly under target and needs urgent attention
- **70–89** → `"warning"` — the hub is below target, so it should be monitored closely
- **≥ 90** → `"normal"` — the hub is performing as expected

This gives us a quick, human-readable label instead of comparing raw numbers every time.

```python
def get_operational_status(efficiency):
    if efficiency < 70:
        return "critical"
    elif 70 <= efficiency < 90:
        return "warning"
    else:
        return "normal"
```

## Hub Data

`delivery_hubs` stores raw delivery data for each hub. Each key is a hub name, and each value is a dictionary holding:

- `actual_deliveries`: packages actually delivered
- `target_deliveries`: packages the hub was expected to deliver

This is the raw input we'll feed into `calculate_efficiency` and `get_operational_status` to evaluate each hub's performance.

```python
delivery_hubs = {
    "Downtown Hub": {
        "actual_deliveries": 180,
        "target_deliveries": 200
    },
    "Westside Hub": {
        "actual_deliveries": 140,
        "target_deliveries": 200
    },
    "Northgate Hub": {
        "actual_deliveries": 195,
        "target_deliveries": 200
    }
}
```

## Evaluating Each Hub

Now we bring everything together. For every hub in our `delivery_hubs` dictionary, we:

1. Pull out `actual_deliveries` and `target_deliveries`
2. Calculate its `efficiency` using `calculate_efficiency`
3. Classify that efficiency into a `status` using `get_operational_status`
4. Print a summary line for that hub

This loop is where the individual functions we built earlier actually get used together to produce a full performance report.

```python
for hub, data in delivery_hubs.items():
    actual = data["actual_deliveries"]
    target = data["target_deliveries"]
    efficiency = calculate_efficiency(actual, target)
    status = get_operational_status(efficiency)

    if status == "critical":
        status_emoji = "🔴"
    elif status == "warning":
        status_emoji = "🟡"
    else:
        status_emoji = "🟢"

    print(f"Hub Name : {hub}")
    print(f"Actual Deliveries  = {actual}")
    print(f"Target Deliveries  = {target}")
    print(f"Efficiency     = {efficiency}%")
    print(f"  {status_emoji} Status         : {status}")
    print()
```
